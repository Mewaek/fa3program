```python
name = input("Enter student's name: ")
score = int(input("Enter score: "))

if score >= 90 and score <= 100:
    grade = "A"
elif score >= 80 and score <= 89:
    grade = "B"
elif score >= 70 and score <= 79:
    grade = "C"
else:
    grade = "F"

print("\nStudent Name:", name)
print("Score:", score)
print("Letter Grade:", grade)
```
